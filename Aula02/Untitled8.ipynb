```python
# Coleta de dados metereológicos

import csv
import requests
import pandas as pd

# URL da API do ThingSpeak
URL = "https://api.thingspeak.com/channels/12397/feeds.json"
Limite = {"results":8000}

resposta = requests.get(URL, params=Limite, timeout=15)
resposta.raise_for_status()

dadosBrutos = resposta.json()

# The 'data' variable is already a dictionary. To create a DataFrame from the feeds, use pd.DataFrame.
df = pd.DataFrame(dadosBrutos['feeds'])
display(df)
```

Os metadados sobre os campos (colunas) estão disponíveis na chave 'channel' do dicionário `dadosBrutos`. Podemos extrair essas descrições para renomear as colunas do DataFrame, tornando-o mais legível.

```python
# Extrair as informações do canal
channel_info = dadosBrutos['channel']

# Criar um dicionário de mapeamento para renomear as colunas
column_rename_map = {}
for i in range(1, 9): # ThingSpeak tem até field8
    field_key = f'field{i}'
    if field_key in channel_info:
        # Limpar o nome da coluna, removendo descrições entre parênteses, se existirem
        cleaned_name = channel_info[field_key].split('(')[0].strip()
        column_rename_map[field_key] = cleaned_name

# Renomear as colunas no DataFrame
df_renamed = df.rename(columns=column_rename_map)

# Exibir as primeiras linhas do DataFrame com as colunas renomeadas
display(df_renamed.head())
```
