```python
students = []

for i in range(5):
    print(f"\nAluno {i+1}:")
    name = input("Digite o nome do aluno: ")
    age = input("Digite a idade do aluno: ")
    students.append({"Nome": name, "Idade": age})

print("\n--- Dados dos Alunos ---")
for i, student in enumerate(students):
    print(f"Aluno {i+1}:")
    print(f"  Nome: {student['Nome']}")
    print(f"  Idade: {student['Idade']}")
```
