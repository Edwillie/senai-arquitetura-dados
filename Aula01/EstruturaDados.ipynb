```python
from collections import deque

class Fila:
    def __init__(self):
        self.items = deque()

    def push(self, item):
        """Adiciona um elemento ao final da fila."""
        self.items.append(item)
        print(f"Adicionado '{item}'. Fila: {list(self.items)}")

    def pop(self):
        """Remove e retorna o elemento da frente da fila."""
        if not self.is_empty():
            removed_item = self.items.popleft()
            print(f"Removido '{removed_item}'. Fila: {list(self.items)}")
            return removed_item
        else:
            print("Erro: A fila está vazia, não é possível remover elementos.")
            return None

    def peek(self):
        """Retorna o elemento da frente da fila sem removê-lo."""
        if not self.is_empty():
            return self.items[0]
        else:
            return None

    def is_empty(self):
        """Verifica se a fila está vazia."""
        return len(self.items) == 0

    def size(self):
        """Retorna o número de elementos na fila."""
        return len(self.items)

# Exemplo de uso da classe Fila
print("\n--- Exemplo de Uso da Classe Fila ---")
minha_fila = Fila()

print("\n--- Inserindo elementos na fila (push) ---")
minha_fila.push("Primeiro")
minha_fila.push("Segundo")
minha_fila.push("Terceiro")

print(f"\nElemento na frente da fila (peek): {minha_fila.peek()}")
print(f"Tamanho da fila: {minha_fila.size()}")

print("\n--- Removendo elementos da fila (pop) ---")
minha_fila.pop()
minha_fila.pop()
minha_fila.pop()

print(f"\nA fila está vazia? {minha_fila.is_empty()}")
minha_fila.pop() # Tentando remover de uma fila vazia
```
